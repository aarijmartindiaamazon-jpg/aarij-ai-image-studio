# Aarij AI Image Studio — Google Colab
Run the cells from top to bottom. Before starting, choose **Runtime → Change runtime type → T4 GPU**. The first model use downloads its weights.

## STEP 1 — Check GPU

```python
import subprocess
result = subprocess.run(['nvidia-smi'], capture_output=True, text=True)
if result.returncode != 0:
    raise RuntimeError('No NVIDIA GPU found. Choose Runtime → Change runtime type → T4 GPU, then reconnect.')
print(result.stdout)
```

## STEP 2 — Mount Google Drive

```python
from google.colab import drive
drive.mount('/content/drive')
from pathlib import Path
OUTPUT_ROOT = Path('/content/drive/MyDrive/Aarij-AI-Image-Studio')
for folder in ['generated-images', 'uploads', 'marketplace', 'lifestyle', 'white-background', 'advertisements', 'temp']:
    (OUTPUT_ROOT / folder).mkdir(parents=True, exist_ok=True)
print(f'Permanent output folder ready: {OUTPUT_ROOT}')
```

## STEP 3 — Get the application
This cell clones the published Aarij AI Image Studio repository automatically. If you already uploaded it to `/content/aarij-ai-image-studio`, the existing folder is used.

```python
import os, subprocess
from pathlib import Path
PROJECT_DIR = Path('/content/aarij-ai-image-studio')
AARIJ_REPO_URL = os.environ.get('AARIJ_REPO_URL', 'https://github.com/aarijmartindiaamazon-jpg/aarij-ai-image-studio.git')
if not (PROJECT_DIR / 'app.py').exists():
    if not AARIJ_REPO_URL:
        raise RuntimeError('Set AARIJ_REPO_URL, or upload the project folder to /content/aarij-ai-image-studio.')
    subprocess.run(['git', 'clone', '--depth', '1', AARIJ_REPO_URL, str(PROJECT_DIR)], check=True)
os.chdir(PROJECT_DIR)
print(f'Project ready: {PROJECT_DIR}')
```

## STEP 4 — Install dependencies

```python
import subprocess, sys
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)
print('Dependencies installed. If Colab reports that a runtime restart is required, restart and Run all again.')
```

## STEP 5 — Configure Hugging Face (optional)

```python
import os
try:
    from google.colab import userdata
    token = userdata.get('HF_TOKEN')
    if token:
        os.environ['HF_TOKEN'] = token
        print('HF_TOKEN loaded securely from Colab Secrets.')
    else:
        print('No HF_TOKEN set. Public models can still download when their terms do not require authentication.')
except Exception:
    print('No HF_TOKEN set. Add it in Colab Secrets only if Hugging Face requests authentication.')
```

## STEP 6 — Launch Aarij AI Image Studio

```python
import os
os.environ['OUTPUT_DIRECTORY'] = '/content/drive/MyDrive/Aarij-AI-Image-Studio'
os.environ['DEFAULT_MODEL'] = 'fast'
os.environ['GRADIO_SHARE'] = 'true'
from ui.gradio_app import build_app
app = build_app()
print('Aarij AI Image Studio is ready. Open the Public URL printed below.')
app.queue(default_concurrency_limit=1).launch(
    share=True,
    show_error=True,
    allowed_paths=['/content/drive/MyDrive/Aarij-AI-Image-Studio'],
)
```
